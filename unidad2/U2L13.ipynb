## 01. Gráficos de líneas simples

Quizá el más sencillo de todos los gráficos es la visualización de una única función y = f(x).

**Componentes básicos de un gráfico**

En Matplotlib, la figura (una instancia de la clase `plt.Figure`) puede considerarse como un único contenedor que incluye todos los objetos que representan ejes, gráficos, texto y etiquetas. Los ejes (una instancia de la clase `plt.Axes`) son un marco delimitador con marcas y etiquetas, que finalmente contendrá los elementos del gráfico que forman nuestra visualización.

Empecemos con una simple sinusoide:

```python
# Generación de los datos
x = np.linspace(0, 10, 1000)  # np.linspace crea un array de 1000 puntos equiespaciados desde 0 hasta 10

# Creación del gráfico
plt.plot(x, np.sin(x))  # Función sinusoidal
plt.plot(x, np.cos(x))  # Función cosenoidal
```

La función `plt.plot()` acepta argumentos adicionales que se pueden usar para personalizar el gráfico.

Del mismo modo, el estilo de línea se puede ajustar mediante el parámetro `linestyle`, con opciones como: 'solid', 'dashed', 'dashdot' y 'dotted'.

```python
# Estilos de línea
plt.plot(x, x + 0, linestyle='solid', label='xxxxx')    # Línea sólida
plt.plot(x, x + 1, linestyle='dashed', label='yyyyy')   # Línea discontinua

plt.legend()           # Añade la leyenda al gráfico
plt.title("Título")    # Título del gráfico
plt.xlabel("Eje x")    # Etiqueta para el eje x
plt.ylabel("Eje y")    # Etiqueta para el eje y
```
