# Numpy

Realizar operaciones matemáticas en Python puede ser complicado, ¿cómo podemos simplificarlo y operar con vectores o incluso matrices?

La respuesta es la librería **Numpy**.

Numpy es una biblioteca de Python que permite trabajar con arrays de manera eficiente. Se utiliza ampliamente en ciencia de datos para tareas como manipulación de datos, filtrado, transformaciones y visualizaciones. A continuación, se muestra un ejemplo básico de cómo usar Numpy para realizar operaciones matemáticas con vectores.

[https://numpy.org/](https://numpy.org/)

Para utilizar Numpy, primero debemos importar la biblioteca. La convención habitual es importarla con el alias `np`.


```python
import numpy as np
```

```python
numeros = [1, 2, 3, 4]  # Creación de una lista de números
```

```python
array = np.array(numeros)  # Creación del array de 1 dimensión a partir de la lista creada anteriormente
```

```python
array  # Comprobación de que los datos dentro del array sean correctos
```

```python
np.arange(0, 20, 2)  # Generación de los números del 0 al 19 de 2 en 2
```

```python
np.random.randint(0, 100)  # Generación de un número aleatorio del 0 al 99
```

```python
array = np.random.randint(0, 100, 10)  # Generación de 10 números aleatorios del 0 al 99
```

```python
array
```

```python
array.max()  # Mostrar el número con el valor más alto
```

```python
array.argmax()  # Mostrar la posición del número con el valor más alto
```

```python
array.min()  # Mostrar el número con el valor más bajo
```

```python
array.mean()  # Mostrar el promedio
```

```python
array[2]  # Obtener el valor en una posición determinada
```

```python
array.reshape(5, 2)  # Mostrar el array en 5 filas y 2 columnas
```

```python
array  # Pero es solo una visualización, no un cambio permanente
```

```python
array = array.reshape(5, 2)  # Hacer el cambio permanente
array
```

```python
array[3, 1]  # Acceder a un elemento en particular del array bidimensional
```

```python
array2 = array[array > 30]  # Filtrar elementos mayores que 30
array2
```

# Tratamiento de valores NaN

**¿Qué es NaN?**

NaN es un valor especial que se utiliza para indicar la ausencia de datos. Dependiendo del tipo de datos, no siempre es posible usar un valor por defecto. Por ejemplo, si dejamos que el usuario introduzca **opcionalmente** una valoración numérica libre, no podemos usar el 0 (ni ningún otro número) para indicar que no ha introducido nad


```python
array = np.asarray([np.nan, 2, np.nan, 3])
array2 = array.copy()

# Sustituir NaN por 0
array2[np.isnan(array2)] = 0
array2
```
