# Pandas

Pandas es una biblioteca de programación en Python que ofrece estructuras de datos y herramientas de manipulación rápidas y flexibles. Está diseñada para facilitar el trabajo con datos "relacionales" o "etiquetados" de forma sencilla e intuitiva.

Pandas es ideal para distintos tipos de datos, desde matrices simples de números hasta series temporales y datos tabulares heterogéneos. La biblioteca proporciona dos estructuras principales: Series y DataFrames.

**¿Por qué es importante para la ciencia de datos?**

La ciencia de datos implica trabajar intensamente con los datos: limpiarlos, transformarlos, analizarlos y visualizarlos.

Por eso, pandas es una pieza clave en el ecosistema de ciencia de datos en Python. Permite importar datos desde múltiples fuentes, realizar una gran variedad de operaciones de manipulación de datos y, además, es compatible con otras bibliotecas de análisis y visualización.

```python
# Pandas, internamente, define cada columna como un vector de Numpy
# Acceso a un campo
df['nombre']
```

```python
# Seleccionar más de un campo
df[['nombre', 'edad']]
```

```python
# Seleccionar solo un elemento de un campo
df['nombre'][3]
```

```python
# Seleccionar un rango de elementos indicando el índice final, pero incluyendo todos los campos del DataFrame. El índice 2 queda excluido.
df[:2]
```

```python
# Seleccionar desde el elemento con índice 2 hasta el final
df[2:]
```

```python
# Seleccionar un elemento desde el índice 2 hasta el 3, excluyendo el 3
df[2:3]
```

```python
# Aplicar filtros al DataFrame para ver campos que cumplen una condición
df['ingresos'] > 2000
```

```python
# Creación de otro DataFrame con solo los elementos que cumplen el filtro
df2 = df[df['ingresos'] > 2000]
df2
```

```python
# Creación de una nueva columna en un DataFrame con edad * ingresos
df2['prueba'] = df2['edad'] * df2['ingresos']
df2
```

```python
# Eliminar una columna de un DataFrame, axis=1 indica que es una columna
df2 = df2.drop('prueba', axis=1)
df2
```

```python
# Eliminar una fila
df2 = df2.drop(index=[2])
df2
```
