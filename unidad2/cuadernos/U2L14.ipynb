# División de los datos en 2 subconjuntos: entrenamiento (70%) y prueba (30%)

```python
# Eliminar solo la columna 'id'
df = df.drop(columns=['id'])

# Separar X (características) e y (etiqueta)
# X = df[['nombre','edad','ingresos']]
X = df.drop(columns=['compra'])  # nombre, edad, ingresos
y = df['compra']
```

```python
# Dividir en train y test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

El **random_state** se utiliza para garantizar la reproducibilidad de la división, y **test_size** determina la proporción de datos que se destinan al conjunto de prueba.

Si no se especifica un valor para **random_state**, Scikit-Learn usará un valor aleatorio, lo que hace que la división sea diferente en cada ejecución. Esto puede ser útil si se desea una división realmente aleatoria, pero dificulta la reproducibilidad en ejecuciones múltiples.

# Dividir en 3 subconjuntos: entrenamiento, validación y prueba

La teoría dice que los datos suelen dividirse en 3 subconjuntos. En entornos reales de desarrollo, lo más habitual es trabajar con solo 2: el de entrenamiento y el de prueba.

Si se desea hacer la división en 3 subconjuntos, la idea sería realizar una primera división, por ejemplo, un 70% para entrenamiento y un 30% para un subconjunto temporal. Luego, ese 30% se dividiría nuevamente en dos partes, por ejemplo, un 15% para validación y un 15% para prueba.
